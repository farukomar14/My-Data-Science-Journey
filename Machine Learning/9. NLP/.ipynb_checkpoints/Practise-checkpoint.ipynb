```python
import nltk
```

```python
sen = "Hi! I love machine learning 89."
```

```python
# Tokenization
from nltk.tokenize import word_tokenize, sent_tokenize
```

```python
token = word_tokenize(sen)
token
```

```python
# Text Cleaning
# Punctuation removal
import string
clean_pun = sen.translate(str.maketrans("","",string.punctuation))
clean_pun
```

```python
clean_digit = clean_pun.translate(str.maketrans("","",string.digits))
clean_digit
```

```python
# Text cleaning - Method 2
```

```python
# Punctuation removal
clean_pun = [word for word in token if word not in string.punctuation]
clean_pun
```

```python
# Digit removal
clean_digit = [word for word in clean_pun if word not in string.digits]
clean_digit
```

```python
clean_text = " ".join(clean_digit)
clean_text
```

```python
# Today
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
import string
from nltk import pos_tag
from nltk.corpus import stopwords
```

```python
sen = "Sure! What would you like to know about? Anything specific on your mind 9, or should I pick a random topic?"
token = word_tokenize(sen)
clean_pun = [word for word in token if word not in string.punctuation]
clean_text = [word for word in clean_pun if word not in string.digits]
pos_word = pos_tag(clean_text)
#pos_word
en_stop_word = set(stopwords.words("English"))
clean_stop_word = [word for word in clean_text if word not in en_stop_word]
clean_stop_word

```
