# 1. Tokenization

```python
from nltk.tokenize import word_tokenize, sent_tokenize
```

```python
token = word_tokenize(sen)
token
```

# 2. Text Cleaning
## Punctuation Removal

```python
import string
```

```python
clean_pun = sen.translate(str.maketrans("","",string.punctuation))
clean_pun
```

```python
# 2nd method of punctuation removal
for word in token:
    if word not in string.punctuation:
        print(word)
```

```python
# 2nd method of punctuation removal 
clean_pun = [word for word in token if word not in string.punctuation] # List Comprehension
clean_pun
```

```python
# Digit Removal
clean_digit = [word for word in clean_pun if word not in string.digits]
clean_digit
```

```python
clean_text = " ".join(clean_digit)
clean_text
```

# 4. StopWord Removal

```python
from nltk.corpus import stopwords
```

```python
en_stop_word = set(stopwords.words("English"))
en_stop_word
```

```python
# Implementation (Step 1 to 4 together)
sen = "Sure! What would you like to know about? Anything specific on your mind 9, or should I pick a random topic?"
# Tokenization
token = word_tokenize(sen)
# punctuation removal
clean_pun = [word for word in token if word not in string.punctuation]
# digit removal
clean_digit = [word for word in clean_pun if word not in string.digits]
# Stopword removal
clean_stop_word = [word for word in clean_digit if word not in en_stop_word]
```

```python
clean_stop_word
```

# 5. Stemming

```python
from nltk.stem import PorterStemmer
ps = PorterStemmer()
```

```python
# wr = ["change","changed","changing","changes","changed"]

for word in clean_stop_word:
    stem_word = ps.stem(word)
    print(stem_word, "<-", word)
```

# 6. Lemmatizer

```python
from nltk.stem import WordNetLemmatizer
lemma = WordNetLemmatizer()
#wr = ["change","changed","changing","changes","changed"]
for word in clean_stop_word:
    lemma_word = lemma.lemmatize(word)
    print(word, "-> ", lemma_word)
    
```

# Vectorizer

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
cv = CountVectorizer()
tf = TfidfVectorizer()
```

```python
count_word = cv.fit_transform(clean_stop_word)
count_word.toarray()
```

```python
tf_word = tf.fit_transform(clean_stop_word)
tf_word.toarray()
```
