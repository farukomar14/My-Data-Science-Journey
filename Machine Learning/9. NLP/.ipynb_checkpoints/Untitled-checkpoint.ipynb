```python
import nltk
import string
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
```

```python
def clean_text(text):
    if pd.isna(text):
        return " "

    text = text.lower()

    rem_pun = text.translate(str.maketrans("","",string.punctuation))
    rem_digit = rem_pun.translate(str.maketrans("","",string.digits))
    words_split = rem_digit.split()

    stop_words = set(stopwords.words('English'))
    rem_stop_words = [word for word in words_split if word not in stop_words]

    lemma = WordNetLemmatizer()
    lemmatized_word = [lemma.lemmatize(words) for words in rem_stop_words]

    pure_text = " ".join(lemmatized_word)

    return pure_text
```

```python
df['Comments'] = df['Comments'].apply(clean_text)
```
