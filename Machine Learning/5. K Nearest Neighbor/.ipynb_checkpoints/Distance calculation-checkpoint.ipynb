```python
dic = {
    'h' : 100.09,
    'g' : 361.04,
    'f' : 4.09,
    'e' : 4.49,
    'd' : 16.16,
    'c' : 64.01,
    'b' : 144.1521,
    'a' : 49.25
}
new = sorted(dic.values())
for n in new:
    #print(n)
    for k in dic.keys():
        if dic[k] == n:
            print(k, n)
```
