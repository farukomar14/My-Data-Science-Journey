```python
df1 = df.copy()
df2 = df.copy()
df3 = df.copy()
```

```python
df1
```

```python
df1['Sex'] = df1['Sex'].replace({'male' : 1, 'female': 0})
```

```python
df1
```

```python
for col in df1.columns:
    if is_numeric_dtype(df1[col]):
        continue
    else:
        uni = df1[col].unique()
    #print(uni)
        if len(uni) == 2:
            df1[col] = df1[col].replace(uni, [0,1])
        if len(uni) == 3:
            df1[col] = df1[col].replace(uni,[0,1,2])
        if len(uni) == 4:
            df1[col] = df1[col].replace(uni,[0,1,2,3])


        
```

```python
df4 = df.copy()
df4
```

```python
df4.drop(['PassengerId','Name','Ticket'],axis = 1,inplace = True)
```

```python
for col in df4.columns:
    if is_numeric_dtype(df4[col]):
        continue
    else:
        dummy = pd.get_dummies(df4[col], drop_first =True, prefix = col).astype(int)
        df4 = df4.drop(col,axis = 1)
        df4 = pd.concat((df4,dummy),axis = 1)
```

```python
df5 = df.copy()
```

```python
df5
```

```python
df5.drop(['PassengerId','Name','Ticket'],axis = 1,inplace = True)
```

```python
import pandas as pd
from pandas.api.types import is_numeric_dtype

categorical_cols = []

for col in df5.columns:
    if not is_numeric_dtype(df5[col]):
        categorical_cols.append(col)

df5 = pd.get_dummies(df5, columns=categorical_cols, drop_first=True, prefix = col).astype(int)
```
