# Feature Selection

## 1. Wrapper Method
### Wrapper methods test different groups of features by training and evaluating a machine learning model on each group Although wrapper methods are more costly in terms of computation than filter methods, they can identify how features work together and often find better feature combinations or the specific algorithm being used

```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.feature_selection import RFE
from sklearn.linear_model import LogisticRegression
```

```python
x = df.drop('payment',axis = 1)
y = df['payment']
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(x,y, test_size = 0.2, random_state=42)
```

```python
lr = LogisticRegression()
lr.fit(xtrain, ytrain)
```

```python
y.value_counts()
```

```python
from imblearn.combine import SMOTETomek
from imblearn.over_sampling import SMOTE
```

```python
sm = SMOTETomek(smote = SMOTE(k_neighbors=2), 
                random_state=100)
newx, newy = sm.fit_resample(x,y)
```

```python
newy.value_counts()
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(newx, newy, test_size = 0.2, random_state=42)
```

```python
lr.fit(xtrain, ytrain)
```

```python
lr.score(xtrain,ytrain)
```

```python
lr.score(xtest,ytest)
```

```python
rfe = RFE(lr, n_features_to_select=10)
rfe
```

```python
rfe.fit(xtrain,ytrain)
```

```python
rfe.score(xtrain,ytrain)*100
```

```python
rfe.score(xtest,ytest)*100
```

```python
sel_fea = xtrain.columns[rfe.support_]
sel_fea
```

```python
rank = rfe.ranking_
rank
```

```python
for feature, ranks in zip(xtrain.columns, rank):
    print(f"{feature} : {ranks}")
```

```python
# Make a separate dataset based on the features which have ranked 1
dic = {}

for feature, ranks in zip(xtrain.columns, rank):
    if ranks == 1:
        print(f'{feature} : {ranks}')
        dic[feature] = ranks

print(dic)
```

```python
# using dictionary comprehension
dic = {
    feature: ranks
    for feature, ranks in zip(xtrain.columns, rank)
    if ranks == 1
}

print(dic)
```

```python
data = pd.DataFrame([dic])
data
```

# 2. Embedded method
Embedded methods combine feature selection with model training. They use
techniques like L1 (Lasso) or L2 (Ridge) regularization to automatically choose the most relevant features
by adding penalties to less important ones.

```python
# Create a Logistic Regression model using L1 regularization (Lasso) with the 'liblinear' solver
lr2 = LogisticRegression(penalty="l1", solver="liblinear")
```

```python
lr2.fit(xtrain,ytrain)
```

```python
coef = lr2.coef_
coef
```

for feature, coe in zip(xtrain.columns, coef[0]):
    print(f"{feature} : {coe}")


# 3. Filter Method
Rank features based on their relevance to the target variable. Common statistical
measures include correlation coefficients, chi square test, information gain, and mutual information
Filter methods are quick and can be used to prepare data before training a model.

```python
from sklearn.ensemble import ExtraTreesClassifier
etc = ExtraTreesClassifier()
etc.fit(xtrain,ytrain)
```

```python
etc.feature_importances_
```

```python
# Make a dataframe
feature_sel = pd.DataFrame(etc.feature_importances_, columns=['Importance'])
feature_sel
```

```python
cols2 = pd.DataFrame(x.columns, columns = ['Columns'])
cols2
```

```python
new = pd.concat([cols2, feature_sel],axis=1)
new
```

```python
new_col = new.nlargest(10,"Importance")
new_col
```

```python
new.plot(kind="bar", color="r")
```

```python
new_col.plot(kind="bar", color="r")
```

# 4. K - best approach

```python
from sklearn.feature_selection import f_classif,chi2, f_regression, SelectKBest
```

```python
kb = SelectKBest(score_func = f_classif, k = 10)
```

```python
feature_score = kb.fit(xtrain,ytrain)
feature_score
```

```python
feature_score.scores_
```

```python
scor = pd.DataFrame(feature_score.scores_,columns=['Feature Score'])
cols3 = pd.DataFrame(xtrain.columns, columns=["Columns"])
```

```python
join = pd.concat([scor, cols3])
join
```

```python
join_lar = join.nlargest(10,"Feature Score")
join_lar
```

```python
plt.figure(figsize=(20,15))
join.plot(kind="bar",color='b')
plt.show()
```

```python
import matplotlib.pyplot as plt

join.plot(kind='bar', color='b', figsize=(10, 8))

plt.title("Selected Features")
plt.xlabel("Features")
plt.ylabel("Rank")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
join_lar.plot(kind='bar',color='g')
```

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# scale first
x_scaled = StandardScaler().fit_transform(newx)

# keep 95% variance
pca = PCA(n_components=0.95)
x_pca = pca.fit_transform(x_scaled)

print("Original shape:", x.shape)
print("Reduced shape:", x_pca.shape)
```
