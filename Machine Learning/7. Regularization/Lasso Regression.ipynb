```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import r2_score
```

```python
df = pd.read_csv('BostonHousing.csv')
df
```

```python
df.shape
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
rm_median = df['rm'].median()
```

```python
df['rm'] = df['rm'].fillna(rm_median)
```

```python
df.isna().sum()
```

```python
df
```

```python
x = df.iloc[:, 0:13]
y = df.iloc[:, -1]
```

```python
y
```

```python
# Feature Scaling
from sklearn.preprocessing import StandardScaler
std = StandardScaler()
x_scaled = std.fit_transform(x)
```

```python
x_scaled
```

```python
# Dataset Split
from sklearn.model_selection import train_test_split as tts
xtrain, xtest, ytrain, ytest = tts(x_scaled, y, test_size=0.2, random_state=42)
```

```python
xtrain.shape
```

```python
xtest.shape
```

```python
ytest.shape
```

```python
ytrain.shape
```

```python
# Apply Linear Regression
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(xtrain, ytrain)
```

```python
lr.score(xtrain, ytrain)
```

```python
lr.score(xtest, ytest)
```

```python
ypred = lr.predict(xtest)
ypred
```

```python
# Model Evaluation
r2_score(ytest, ypred)
```

```python
# Apply Lasso Regression
from sklearn.linear_model import Lasso
ls = Lasso(alpha=0.1)
```

```python
ls.fit(xtrain, ytrain)
```

```python
ls.score(xtrain, ytrain)
```

```python
ls.score(xtest, ytest)
```

```python
ls_pred = ls.predict(xtest)
ls_pred
```

```python
r2_score(ytest, ls_pred)
```

```python
import numpy as np
ls = Lasso()
para = {'alpha' : np.logspace(-1, 10, 1000)}
para
```

```python
# GridSearchCV
from sklearn.model_selection import GridSearchCV
gsc = GridSearchCV (ls, para, cv = 8, scoring="neg_mean_absolute_error")
gsc.fit(xtrain, ytrain)
```

```python
best_para = gsc.best_params_
best_para
```

```python
gsc.score(xtrain, ytrain)
```

```python
gsc.score(xtest, ytest)
```

```python
gsc_pred = gsc.predict(xtest)
gsc_pred
```

```python
r2_score(ytest, gsc_pred)
```
