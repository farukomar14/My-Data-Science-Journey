# Visualization

```python
ax = sns.countplot(x='region', hue='region', data=df)

# Annotate each bar
for p in ax.patches:
    height = p.get_height()
    ax.annotate(f'{height:.0f}', 
                (p.get_x() + p.get_width()/2., height),
                ha='center', va='bottom')
plt.legend()
plt.show()
```

```python
df.smoker.value_counts()
```

```python
# Smoker distribution across regions

# Plot barplot
ax = sns.countplot(data =df, x='smoker', hue = df.region)

# Annotate each bar
for p in ax.patches:
    height = p.get_height()
    ax.annotate(f'{height:.0f}',
                (p.get_x() + p.get_width()/2, height),
                ha='center', va='bottom')

plt.show()
```

```python
df.head(100)
```

```python


# Aggregate sum of smokers per region and children
grouped_sum = df.groupby(['region','children'])['smoker'].sum().reset_index()

# Plot barplot
ax = sns.barplot(x='region', y='smoker', hue='children', data=grouped_sum, palette='viridis')

# Annotate each bar
for p in ax.patches:
    height = p.get_height()
    ax.annotate(f'{height:.0f}',
                (p.get_x() + p.get_width()/2, height),
                ha='center', va='bottom')

plt.show()
```

```python
df.children.value_counts()
```

```python
df.groupby('region')['children'].sum()
```

```python
sns.histplot(df.bmi, kde=True, bins=30, edgecolor='black')
plt.title('BMI Distribution')
plt.xlabel('BMI')
plt.ylabel('Count')
plt.show()
```

```python
# Aggregate mean BMI per smoker status and region
bmi_grouped = df.groupby(['smoker','region'])['bmi'].mean().reset_index()

# Plot barplot
ax = sns.barplot(x='smoker', 
                 y='bmi', 
                 hue='region', 
                 data=bmi_grouped, 
                 palette='Set1')

# Annotate each bar
for p in ax.patches:
    height = p.get_height()
    ax.annotate(f'{height:.1f}',  # 1 decimal for BMI
                (p.get_x() + p.get_width()/2, height),
                ha='center', va='bottom')

plt.title('BMI per smoker status and region')
plt.show()
```

# Encoding

## Label Encoding (Using Library )

```python
df.head()
```

```python
df1 = df.copy()
df2 = df.copy()
df3 = df.copy()
df4 = df.copy()
df5 = df.copy()
```

```python
df1.head()
```

```python
df1.info()
```

```python
from pandas.api.types import is_numeric_dtype
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
le = LabelEncoder()
ohe = OneHotEncoder()
```

```python
df1['sex'] = le.fit_transform(df1['sex'])
df1.head()
```

```python
df1['smoker'] = le.fit_transform(df1['smoker'])
df1.head()
```

```python
df1['region'] = le.fit_transform(df1['region'])
df1.head()
```

```python
df1
```

### df4

```python
df4.head()
```

```python
df4.columns
```

```python
for col in df4.columns:
    if is_numeric_dtype(df4[col]):
        continue
    else:
        df4[col] = le.fit_transform(df4[col])
df4.head()
```

## Label Encoding (Manually -> df2)

```python
df2['sex'] = df2['sex'].replace({'female': 0, 'male': 1})
df2.head()
```

```python
df2['smoker'] = df2['smoker'].replace({'no': 0, 'yes': 1})
df2.head()
```

### df3

```python
df3.columns
```

```python
for col in df3.columns:
    if is_numeric_dtype(df3[col]):
        continue
    uni = df3[col].unique()
    #print(uni)
    #print(len(uni))
    if df3[col].nunique() == 2:
        df3[col] = df3[col].replace(uni, [0,1])
    if df3[col].nunique() == 3:
        df3[col] = df3[col].replace(uni, [0,1,2])
    if df3[col].nunique() == 4:
        df3[col] = df3[col].replace(uni, [0,1,2,3])
```
