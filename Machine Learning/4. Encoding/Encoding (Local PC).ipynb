```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
from pandas.api.types import is_numeric_dtype
```

```python
df = pd.read_csv('insurance.csv')
df
```

```python
df1 = df.copy()
df2 = df.copy()
df3 = df.copy()
df4 = df.copy()
df5 = df.copy()
```

# Manually Encoding -> Label Encoding

```python
df1.shape
```

```python
df.info()
```

```python
df1.columns
```

```python
for col in df1.columns:
    if is_numeric_dtype(df1[col]):
        continue
    else:
        uni = df1[col].unique()
        #print(uni)
        if len(uni) == 2:
            df1[col] = df1[col].replace(uni, [0,1])
        if len(uni) == 3:
            df1[col] = df1.col.replace(uni, [0,1,2])
        if len(uni) == 4:
            df1[col] == df1[col].replace(uni,[0,1,2,3])

        
```

```python
df1
```

# Label Encoding -> Using scikit learn library

```python
df2.head()
```

```python
df2.columns
```

```python
le = LabelEncoder()
for col in df2.columns:
    if is_numeric_dtype(df2[col]):
        continue
    else:
        df2[col] = le.fit_transform(df2[col])
        
```
